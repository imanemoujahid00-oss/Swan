```python
quality_report = {
    "dataset_name": "moisturizers_with_cosing.csv",

    "number_of_rows": int(df.shape[0]),

    "number_of_columns": int(df.shape[1]),

    "column_names": df.columns.tolist(),

    "data_types": {
        col: str(dtype)
        for col, dtype in df.dtypes.items()
    },

    "missing_values": {
        col: int(count)
        for col, count in df.isnull().sum().items()
    },

    "missing_percentage": {
        col: round(float(value), 2)
        for col, value in (
            df.isnull().mean() * 100
        ).items()
    },

    "exact_duplicate_rows": int(
        df.duplicated().sum()
    ),

    "report_created_at": (
        datetime.now(timezone.utc).isoformat()
    )
}
```

```python
docs_folder = project_dir / "docs"

docs_folder.mkdir(
    parents=True,
    exist_ok=True
)

report_path = (
    docs_folder / "initial_data_report.json"
)

with open(report_path, "w", encoding="utf-8") as f:
    json.dump(
        quality_report,
        f,
        indent=4,
        ensure_ascii=False
    )

print("Data quality report saved successfully!")
print("Location:", report_path)
```

```python
with open(report_path, "r", encoding="utf-8") as f:
    saved_report = json.load(f)

print("Products:", saved_report["number_of_rows"])
print("Columns:", saved_report["number_of_columns"])
print(
    "Exact duplicates:",
    saved_report["exact_duplicate_rows"]
)
```

```python
import hashlib

def calculate_sha256(path):
    sha256 = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(
            lambda: f.read(8192),
            b""
        ):
            sha256.update(chunk)

    return sha256.hexdigest()

file_hash = calculate_sha256("/Users/linanawwas/code/imanemoujahid00-oss/Swan/skincare_project/data/raw/moisturizers_with_cosing.csv")

print("SHA-256 fingerprint:")
print(file_hash)
```

```python
quality_report["sha256"] = file_hash

with open(report_path, "w", encoding="utf-8") as f:
    json.dump(
        quality_report,
        f,
        indent=4,
        ensure_ascii=False
    )
```
